```python
def create_rolling_feature(df,source,column,window):
    feature=df.groupby(source)[column].transform(
        lambda s:s.shift(1).rolling(window,min_periods=1).mean()
    )
    return feature
```

```python
def rolling_HW_goals_scored(df,window=5):
    df = df.sort_values("MatchDate").copy()
    df["HomeGoalsMeanL5"] = (
        df.groupby("HomeTeam")["FullTimeHomeGoals"]
          .transform(lambda x: x.shift(1).rolling(window, min_periods=1).mean())
    )

    df["AwayGoalsMeanL5"] = (
        df.groupby("AwayTeam")["FullTimeAwayGoals"]
          .transform(lambda x: x.shift(1).rolling(window, min_periods=1).mean())
    )

    return df
```

```python
def rolling_HW_goals_conceded(df,window=5):
    df=df.sort_values('MatchDate').copy()
    df['HomeGoalsConcededL5']=(df.groupby('HomeTeam')['FullTimeAwayGoals']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
    df['AwayGoalsConcededL5']=(df.groupby('AwayTeam')['FullTimeHomeGoals']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
    return df
```

```python
import os

all_odds_list = []

for season, filepath in season_to_file.items():
    if os.path.exists(filepath):
        # 1. Read the season file
        temp_df = pd.read_csv(filepath,encoding='latin1',on_bad_lines='skip')
        
        # Ensure core identification columns are present
        if set(['Date', 'HomeTeam', 'AwayTeam']).issubset(temp_df.columns):
            
            # 2. Standardize dates
            temp_df['Date'] = pd.to_datetime(temp_df['Date'], dayfirst=True)
            
            # 3. Initialize combined odds columns with NaN
            temp_df['HomeOdds'] = np.nan
            temp_df['DrawOdds'] = np.nan
            temp_df['AwayOdds'] = np.nan
            
            # 4. First priority: Bet365
            if 'B365H' in temp_df.columns:
                temp_df['HomeOdds'] = temp_df['HomeOdds'].fillna(temp_df['B365H'])
            if 'B365D' in temp_df.columns:
                temp_df['DrawOdds'] = temp_df['DrawOdds'].fillna(temp_df['B365D'])
            if 'B365A' in temp_df.columns:
                temp_df['AwayOdds'] = temp_df['AwayOdds'].fillna(temp_df['B365A'])
                
            # 5. Second priority: Interwetten (Fallback for missing/NaN values)
            if 'IWH' in temp_df.columns:
                temp_df['HomeOdds'] = temp_df['HomeOdds'].fillna(temp_df['IWH'])
            if 'IWD' in temp_df.columns:
                temp_df['DrawOdds'] = temp_df['DrawOdds'].fillna(temp_df['IWD'])
            if 'IWA' in temp_df.columns:
                temp_df['AwayOdds'] = temp_df['AwayOdds'].fillna(temp_df['IWA'])
            
            # Keep only standard key columns + consolidated odds
            clean_odds = temp_df[['Date', 'HomeTeam', 'AwayTeam', 'HomeOdds', 'DrawOdds', 'AwayOdds']].copy()
            all_odds_list.append(clean_odds)
            
    else:
        print(f"File missing: {filepath}")

# Combine all seasons into a single lookup table
odds_df = pd.concat(all_odds_list, ignore_index=True)
odds_df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
import pandas as pd


class FootballPreprocessor:

    """
    Football preprocessing pipeline.

    Responsibilities
    ----------------
    • Team-aware missing value imputation
    • Optional categorical encoding
    • Optional feature scaling
    • Preserve feature ordering
    """

    #########################################################
    # Feature groups
    #########################################################

    HOME_FEATURES = [

        "HomeGoalsMeanL5Venuedpd",
        "HomeGoalsConcededL5Venuedpd",
        "HomeGoalDFMeanL5Venuedpd",
        "HomePointsL5Venuedpd",

        "HomeGoalsConcededL5",
        "HomeGoalsMeanL5",
        "HomeShotsOTMeanL10",
        "HomeGoalDFMeanL10",
        "HomeGoalDFMeanL5",
        "HomePointsMeanL5",
        "HomePointsMeanL10",
        "HomeGoalsMeanL10",
        "HomeGoalsConcededL10",
        "Homewin_streak",
        "HomeH2HL5",
        "HomeRedCardL2",
        'Homeodds'
    ]

    AWAY_FEATURES = [

        "AwayGoalsMeanL5Venuedpd",
        "AwayGoalsConcededL5Venuedpd",
        "AwayGoalDFMeanL5Venuedpd",
        "AwayPointsL5Venuedpd",

        "AwayGoalsConcededL5",
        "AwayGoalsMeanL5",
        "AwayShotsOTMeanL10",
        "AwayGoalDFMeanL10",
        "AwayGoalDFMeanL5",
        "AwayPointsMeanL5",
        "AwayPointsMeanL10",
        "AwayGoalsMeanL10",
        "AwayGoalsConcededL10",
        "Awaywin_streak",
        "AwayH2HL5",
        "AwayRedCardL2"

    ]
    MARKET_FEATURES = [
    "HomeOdds",
    "DrawOdds",
    "AwayOdds",
    "HomeProb",
    "DrawProb",
    "AwayProb",
    "MatchBalance",
]

    #########################################################

    def __init__(self,
                 encode=False,
                 scale=False):

        self.encode = encode
        self.scale = scale

        self.numeric_cols = None
        self.categorical_cols = None
        self.feature_order = None

        self.global_medians = None
        self.home_team_medians = None
        self.away_team_medians = None

        if encode:

            self.encoder = OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1
            )

        else:

            self.encoder = None

        if scale:

            self.scaler = StandardScaler()

        else:

            self.scaler = None

    #########################################################
    # FIT
    #########################################################

    def fit(self, X):

        X = X.copy()

        self.feature_order = X.columns.tolist()

        self.numeric_cols = X.select_dtypes(
            include="number"
        ).columns.tolist()

        self.categorical_cols = X.select_dtypes(
            exclude="number"
        ).columns.tolist()

        #####################################################

        self.global_medians = X[self.numeric_cols].median()

        #####################################################

        home_cols = [
            c for c in self.HOME_FEATURES
            if c in X.columns
        ]

        away_cols = [
            c for c in self.AWAY_FEATURES
            if c in X.columns
        ]

        self.home_team_medians = (
            X.groupby("HomeTeam")[home_cols]
            .median()
        )

        self.away_team_medians = (
            X.groupby("AwayTeam")[away_cols]
            .median()
        )

        #####################################################

        if self.encode:

            self.encoder.fit(
                X[self.categorical_cols]
            )

        #####################################################

        if self.scale:

            X = self._impute(X)

            self.scaler.fit(
                X[self.numeric_cols]
            )

        return self

    #########################################################
    # IMPUTATION
    #########################################################

    def _impute(self, X):

        X = X.copy()

        #####################################################
        # HOME FEATURES
        #####################################################

        for col in self.HOME_FEATURES:

            if col not in X.columns:
                continue

            X[col] = (
                X[col]
                .fillna(
                    X["HomeTeam"].map(
                        self.home_team_medians[col]
                    )
                )
                .fillna(
                    self.global_medians[col]
                )
            )

        #####################################################
        # AWAY FEATURES
        #####################################################

        for col in self.AWAY_FEATURES:

            if col not in X.columns:
                continue

            X[col] = (
                X[col]
                .fillna(
                    X["AwayTeam"].map(
                        self.away_team_medians[col]
                    )
                )
                .fillna(
                    self.global_medians[col]
                )
            )
        #####################################################
        # MARKET FEATURES
        #####################################################
        for col in self.MARKET_FEATURES:
            if col not in X.columns:
                continue
            X[col] =(
                X[col].fillna(
                    self.global_medians[col]
                )
            )

        return X
    #########################################################
    # TRANSFORM
    #########################################################

    def transform(self, X):

        X = X.copy()

        X = self._impute(X)

        if self.encode:

            X[self.categorical_cols] = self.encoder.transform(
                X[self.categorical_cols]
            )

        if self.scale:

            X[self.numeric_cols] = self.scaler.transform(
                X[self.numeric_cols]
            )

        return X[self.feature_order]

    #########################################################
    # FIT + TRANSFORM
    #########################################################

    def fit_transform(self, X):

        self.fit(X)

        return self.transform(X)
```
