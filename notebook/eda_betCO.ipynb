```python
home_df = pd.DataFrame({
    "MatchDate": df1["MatchDate"],
    "Season": df1["Season"],

    "Team": df1["HomeTeam"],
    "Opponent": df1["AwayTeam"],

    "Venue": "Home",
    "Shots":df1['HomeShotsOnTarget'],
    "GoalsFor": df1["FullTimeHomeGoals"],
    "GoalsAgainst": df1["FullTimeAwayGoals"],

    "Result": df1["FullTimeResult"],
    "RedCard": df1['HomeRedCards']
})
away_df=pd.DataFrame({
    "MatchDate": df1["MatchDate"],
    "Season": df1["Season"],

    "Team": df1["AwayTeam"],
    "Opponent": df1["HomeTeam"],

    "Venue": "Away",
    "Shots":df1['AwayShotsOnTarget'],
    "GoalsFor": df1["FullTimeAwayGoals"],
    "GoalsAgainst": df1["FullTimeHomeGoals"],
    "Result": df1["FullTimeResult"],
    "RedCard": df1['AwayRedCards']
})
```

```python
away_df.sample(10)
```

```python
team_df = pd.concat(
    [home_df, away_df],
    ignore_index=True
)
```

```python
team_df = (
    team_df
    .sort_values(["Team", "MatchDate"])
    .reset_index(drop=True)
)
team_df.head()
```

```python
team_df["GoalsConcededL5"] = (
    team_df.groupby("Team")["GoalsAgainst"]
           .transform(
               lambda s: s.shift(1).rolling(5, min_periods=1).mean()
           )
)
```

```python
team_df["GoalsMeanL5"] = (
    team_df.groupby("Team")["GoalsFor"]
           .transform(
               lambda s: s.shift(1).rolling(5, min_periods=1).mean()
           )
)
```

```python
team_df["ShotsOTMeanL5"] = (
    team_df.groupby("Team")["Shots"]
           .transform(
               lambda s: s.shift(1).rolling(5, min_periods=1).mean()
           )
)
```

```python
team_df.sample(10)
```

```python
def get_points(df):
    if (df['Venue']=='Home' and df['Result']=='H') or (df['Venue']=='Away' and df['Result']=='A'):
        point=3
    elif df['Result']=='D':
        point=1
    else:
        point=0
    return point
    
```

```python
team_df['points']=team_df.apply(get_points,axis=1)
team_df.head()
```

```python
team_df["PointsMeanL5"] = (
    team_df.groupby("Team")["points"]
           .transform(
               lambda s: s.shift(1).rolling(5, min_periods=1).mean()
           )
)
team_df.head()
```

```python
def create_rolling_feature(df,source,column,window):
    feature=df.groupby(source)[column].transform(
        lambda s:s.shift(1).rolling(window,min_periods=1).mean()
    )
    return feature
```

```python
team_df['goal_df']=team_df.GoalsFor - team_df.GoalsAgainst
team_df['GoalDFMeanL5']=create_rolling_feature(team_df,'Team','goal_df',5)
team_df['RedCardL2']= create_rolling_feature(team_df,'Team','RedCard',2)
team_df.head()
```

```python
team_df['GoalDFMeanL10']=create_rolling_feature(team_df,'Team','goal_df',10)
team_df['ShotsOTMeanL10']=create_rolling_feature(team_df,'Team','Shots',10)
team_df['PointsMeanL10']=create_rolling_feature(team_df,'Team','points',10)
team_df.head()
```

```python
team_df.sample(10)
```

```python
team_df['GoalsMeanL10']=create_rolling_feature(team_df,'Team','GoalsFor',10)
team_df['GoalsConcededL10']=create_rolling_feature(team_df,'Team','GoalsAgainst',10)
```

```python
result_map={
    0:'L',
    1:'D',
    3:'W'
}
team_df['outcome']=team_df['points'].map(result_map)
team_df.head()
```

```python
def calc_win_streak(results):
    streak=0
    out=[]
    for result in results:
        out.append(streak)
        if result=='W':
            streak+=1
        elif result=='L' or result=='D':
            streak=0
    return out
```

```python
team_df['win_streak']=team_df.groupby('Team')['outcome'].transform(calc_win_streak)
team_df.head()
```

```python
def rolling_HW_goals_scored(df,window=5):
    df = df.sort_values("MatchDate").copy()
    df["HomeGoalsMeanL5"] = (
        df.groupby("HomeTeam")["FullTimeHomeGoals"]
          .transform(lambda x: x.shift(1).rolling(window, min_periods=1).mean())
    )

    df["AwayGoalsMeanL5"] = (
        df.groupby("AwayTeam")["FullTimeAwayGoals"]
          .transform(lambda x: x.shift(1).rolling(window, min_periods=1).mean())
    )

    return df
```

```python
def rolling_HW_goals_conceded(df,window=5):
    df=df.sort_values('MatchDate').copy()
    df['HomeGoalsConcededL5']=(df.groupby('HomeTeam')['FullTimeAwayGoals']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
    df['AwayGoalsConcededL5']=(df.groupby('AwayTeam')['FullTimeHomeGoals']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
    return df
```

```python
def rolling_HW_shots_onTarget(df,window=5):
    df=df.sort_values('MatchDate').copy()
    df['HomeShotTarget']=(df.groupby('HomeTeam')['HomeShotsOnTarget']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
    df['AwayShotTarget']=(df.groupby('AwayTeam')['AwayShotsOnTarget']
        .transform(lambda x:x.shift(1).rolling(window,min_periods=1).mean())
                            )
```
